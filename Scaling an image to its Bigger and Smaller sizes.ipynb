```python
img = cv2.imread('ostrich.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.title("Original Image")
plt.axis('off')
plt.show()
```

```python
# Bigger image (2x)
bigger = cv2.resize(img, None, fx=2, fy=2, interpolation=cv2.INTER_LINEAR)

# Smaller image (0.5x)
smaller = cv2.resize(img, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_AREA)
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(bigger)
plt.title("Bigger (2x)")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(smaller)
plt.title("Smaller (0.5x)")
plt.axis('off')

plt.show()
```
