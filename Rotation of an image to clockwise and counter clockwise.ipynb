```python
img = cv2.imread('zebra.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.title("Original Image")
plt.axis('off')
plt.show()
```

```python
clockwise = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

plt.imshow(clockwise)
plt.title("90° Clockwise Rotation")
plt.axis('off')
plt.show()
```

```python
counter = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)

plt.imshow(counter)
plt.title("90° Counter Clockwise Rotation")
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(clockwise)
plt.title("Clockwise")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(counter)
plt.title("Counter Clockwise")
plt.axis('off')

plt.show()
```
