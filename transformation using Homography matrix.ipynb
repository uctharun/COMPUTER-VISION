```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('panda (1).png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image)
plt.axis('off')
plt.show()
```

```python
h, w = image.shape[:2]

src_points = np.float32([
    [0, 0],
    [w-1, 0],
    [w-1, h-1],
    [0, h-1]
])

dst_points = np.float32([
    [50, 50],
    [w-100, 20],
    [w-50, h-50],
    [20, h-20]
])
```

```python
H, status = cv2.findHomography(src_points, dst_points)
print("Homography Matrix:")
print(H)
```

```python
result = cv2.warpPerspective(image, H, (w, h))
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(result)
plt.title("Homography Transformed Image")
plt.axis('off')

plt.show()
```
