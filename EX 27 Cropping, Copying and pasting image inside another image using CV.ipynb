```python
import cv2
import matplotlib.pyplot as plt
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
print(uploaded.keys())
```

```python
import os

filenames = list(uploaded.keys())

image1 = cv2.imread(filenames[0])
image2 = cv2.imread(filenames[1])

print("Image 1:", filenames[0], image1.shape)
print("Image 2:", filenames[1], image2.shape)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image1, cv2.COLOR_BGR2RGB))
plt.title("Main Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.title("Image to Crop")
plt.axis("off")

plt.show()
```

```python
crop = image2[50:250, 50:250]
```

```python
plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.title("Cropped Image")
plt.axis("off")
plt.show()
```

```python
crop = cv2.resize(crop, (200, 200))
```

```python
result = image1.copy()

x = 50
y = 50

result[y:y+200, x:x+200] = crop
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image1, cv2.COLOR_BGR2RGB))
plt.title("Original Main Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.title("Cropped Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title("Cropped Image Pasted")
plt.axis("off")

plt.show()
```
